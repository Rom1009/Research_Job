````python
client = Groq()

class Validation(BaseModel):
  is_valid: bool
  syntax_error: list[str]

class SchemaCVResponse(BaseModel):
  skills: list[str]
  education: list[str]
  work_experience: list[str]

  additional_info: list[str]

  validation_status: Validation
````

````python
data_md_list = glob.glob(PATH_DATA + "**.md")

for data_md in data_md_list:
    with open(data_md, "r", encoding = "utf-8") as f:
        markdown_data = f.read()

    response = client.chat.completions.create(
        model = MODEL_NAME,
        messages = [
            {
                "role": "system",
                "content": "You are a CV checker expert. Validate the CV carefully and syntax validation and metadata."
            },
            {
                "role": "user",
                "content": f"Base on the {markdown_data}. Extract the skills, educations, work_experience, additional_info and validate syntax of format"
            }
        ],
        response_format = {
            "type": "json_schema",
            "json_schema": {
                "name": "SchemaCVResponse",
                "schema": SchemaCVResponse.model_json_schema()
            }
        }
    )

    scan = SchemaCVResponse.model_validate(json.loads(response.choices[0].message.content))

    output_dir = os.path.join(PATH_DATA, data_md.split("/")[1].split(".")[0]) + ".json"
    with open(output_dir, "w", encoding = "utf-8") as f:
        json.dump(scan.model_dump(), f, indent = 2)
````

````python
import asyncio
import httpx
from bs4 import BeautifulSoup
import json
import re
import random

from fake_useragent import UserAgent

async def get_job_description(client, job_url: str) -> str:
    try:
        match = re.search(r'-(\d+)(?:/|\?|$)', job_url) or re.search(r'/view/{\d+}', job_url)
        if not match:
            return "Can not find the job_id"

        job_id = match.group(1)
        detail_api_url = f"https://www.linkedin.com/jobs-guest/jobs/api/jobPosting/{job_id}"

        response = await client.get(detail_api_url)
        if response.status_code == 200:
            soup = BeautifulSoup(response.text, "html.parser")

            jd_tag = soup.find("div", class_= "description__text")
            if jd_tag:
                return jd_tag.get_text(separator = "\n", strip = False)

        return f"Can not download {response.status_code}"

    except Exception as e:
        print(f"Wrong: {e}")
````

````python
async def scrape_linkedin_job(keywords: str, location: str, filter_level: str, max_pages: int = 1):
    base_search_url = "https://www.linkedin.com/jobs-guest/jobs/api/seeMoreJobPostings/search"
    ua = UserAgent()

    headers = {
        "User-Agent": ua.random,
        "Accept-Language": "en-US,en;q=0.9",
        "Referer": "https://www.linkedin.com/"
    }

    all_jobs = []

    async with httpx.AsyncClient(headers = headers, timeout = 20.0) as client:
        for page in range(max_pages):
            start_idx = page * 25

            params = {
                "keywords": keywords,
                "location": location,
                "start": start_idx,
                "f_E":  filter_level
            }
            print(f"\n🔄 [Trang {page + 1}] Đang tìm kiếm việc làm...")

            try:
                response = await client.get(base_search_url, params = params)

                if response.status_code != 200:
                    print(f"Stop scraping: {response.status_code}")
                    break

                soup = BeautifulSoup(response.text, "html.parser")
                job_cards = soup.find_all("li")

                if not job_cards:
                    print("Completed")
                    break

                print(f"Finding {len(job_cards)}")

                for card in job_cards:
                  title_tag = card.find("h3", class_ = "base-search-card__title")
                  company_tag = card.find("h4", class_ = "base-search-card__subtitle")
                  location_tag = card.find("span", class_ = "job-search-card__location")
                  link_tag = card.find("a", class_ = "base-card__full-link")

                  if title_tag and company_tag and link_tag:
                      job_url = link_tag["href"]
                      print(f"URL: {job_url}")
                      print(f" Reading JD: {title_tag.get_text(strip = False)} ")

                      job_description = await get_job_description(client, job_url)

                      job_data = {
                          "title": title_tag.get_text(strip = True),
                          "company": company_tag.get_text(strip = True),
                          "location": location_tag.get_text(strip = True) if location_tag else "No clear",
                          "url": job_url,
                          "description": job_description
                      }
                      all_jobs.append(job_data)

                      await asyncio.sleep(random.uniform(1.5, 3.5))

                await asyncio.sleep(2)

            except Exception as e:
                print(f"Error: {e}")
                break

    return all_jobs
````

````python
async def main():

    keyword_search = "AI Engineer"
    location_search = "Vietnam"
    page_to_scrape = 1
    filter_level = "1,2"

    print(f"Start scraping Linkedin")
    print(f"Keyword: {keyword_search} | Location: {location_search}")

    results = await scrape_linkedin_job(keywords = keyword_search, location = location_search, filter_level = filter_level, max_pages = page_to_scrape)

    print(f"Completed {len(results)}")
    if results:
        print("\n This is the answer: ")
        print(json.dumps(results[0], indent=4, ensure_ascii=False))

        with open("linkedin_jobs.json", "w", encoding="utf-8") as f:
            json.dump(results, f, indent=4, ensure_ascii=False)
````

````python
await main()
````

````python
class ScoreCV(BaseModel):
    skill_score: int
    education_score: int
    work_experience_score: int
    project_score: int

    matched_skills: list[str]

    gap_analysis: list[str]
    actionable_advice: list[str]
    evaluation_summary: str

    project_impact: list[str]
    technical_complexity: list[str]
````

````python
client = Groq()
data_json_list = glob.glob(PATH_DATA + "**.json")

for i in range(len(linkedin_jobs)):
    with open("/home/thomas/Desktop/AI/Research_Job/docs/data.json", "r", encoding = "utf-8") as f:
        json_data = json.load(f)

    response = client.chat.completions.create(
        model = MODEL_NAME,
        temperature = 0.0,
        messages = [
            {
                "role": "system",
                "content":
                '''
                  You are a strict and objective Technical Recruiter expert. Your task is to match a candidate's CV (provided as JSON) against a Job Description (JD).
                  CRITICAL RULES:
                  1. You must ONLY return a valid JSON object. Do not include any markdown formatting wrappers like ```json, no introductory text, and no concluding remarks.
                  2. Every judgment in 'gap_analysis' and 'matched_skills' must be based strictly on the provided text. Do not assume or hallucinate skills.
                  3. In 'evaluation_summary', provide a professional, holistic critique from a human recruiter's perspective. Balance the strict technical scores with an assessment of the candidate's potential, career progression, and how well their overall background fits the company's context.
                  4. In 'technical_complexity', assess whether the projects demonstrate advanced software engineering (such as system design, optimization, security, or data handling) or are just basic tutorial-level apps.
                  5. In 'project_impact', look for concrete problem-solving evidence, architectural decisions (why a technology was chosen), or any metrics and results mentioned in the README.
                  6. In 'skill_score': Evaluate on a scale from 0 to 100 based on the percentage of mandatory technical keywords in the JD that are present in the CV.
                  7. In 'education_score': Evaluate on a scale from 0 to 100 based on the relevance of the candidate's degree, major, and professional certifications to the role.
                  8. In 'work_experience_score': Evaluate on a scale from 0 to 100 based on the years of relevant experience, job titles, and career progression matching the JD requirements.
                  9. In 'project_score', score on a scale from 0 to 100 based on the repository's README complexity, architectural decisions, and evidence of practical problem-solving.
                '''
            },
            {
                "role": "user",
                "content": f"""
                Please evaluate how well the candidate's CV with work experience, skills, education and their practical projects matches the Job Description (JD) provided below.

                📝 [CANDIDATE CV with work experience (JSON)]
                {json_data["work_experience"]}
                📝 [CANDIDATE CV with skill (JSON)]
                {json_data["skills"]}
                📝 [CANDIDATE CV with education (JSON)]
                {json_data["education"]}

                💻 [PRACTICAL PROJECT EVIDENCE (README)]
                {readme_markdown}

                🎯 [JOB DESCRIPTION (JD)]
                {linkedin_jobs[i]["description"]}

                ⚙️ [REQUIRED OUTPUT FORMAT]
                You must output a single JSON object with this exact structure:
                {{
                    "skill_score": <integer between 0 and 100 based on CANDIDATE CV with skill>,
                    "education_score": <integer between 0 and 100 CANDIDATE CV with education>,
                    "work_experience_score": <integer between 0 and 100 CANDIDATE CV with work experience>,
                    "project_score": <integer between 0 and 100 based on PRACTICAL PROJECT>
                    "matched_skills": [<list of skills from CV that match the JD>],
                    "gap_analysis": [<list of critical requirements or skills in JD missing from CV>],
                    "actionable_advice": [<list of specific recommendations to improve the CV for this job>]
                    "evaluation_summary": <a professional, holistic critique from a human recruiter's perspective>
                    "project_impact": [<list of project impact assessment>]
                    "technical_complexity": [<list of technical complexity assessment>]
                }}
                """
            }
        ],
        response_format = {
            "type": "json_schema",
            "json_schema": {
                "name": "ScoreCV",
                "schema": ScoreCV.model_json_schema()
            }
        }
    )

    scan = ScoreCV.model_validate(json.loads(response.choices[0].message.content))

    output_dir = f"matching/matching{i}.json"
    with open(output_dir, "w", encoding = "utf-8") as f:
        json.dump(scan.model_dump(), f, indent = 2)

````

````python
matching_lists = glob.glob("matching/**.json")
````

````python
results = []
for matching in matching_lists:
    with open(matching, "r", encoding = "utf-8") as f:
        matching_data = json.load(f)

    results.append(matching_data)
````

````python
for res in results:
    res["match_percentage"] = res["skill_score"] * 0.4 + res["education_score"] * 0.15 + \
                              res["work_experience_score"] * 0.05 + res["project_score"] * 0.4
````

````python
top_score = sorted(results, key = lambda x: x["match_percentage"], reverse = True)
````
